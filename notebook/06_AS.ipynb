### unclassified Ruminococcaceae 

Resistant Starch (RS) acts as a critical fuel source for the gut microbiome, but its health benefits depend entirely on a specialized group of bacteria known as primary degraders. These microbes perform the essential final step of fermentation, converting RS into Short-Chain Fatty Acids (SCFAs)—specifically acetate, propionate, and butyrate. These metabolites play a vital role in neuroimmune-endocrine modulation, serving as a chemical bridge between the gut and the rest of the body to prevent metabolic and inflammatory diseases.

However, this system is fragile: if these specific degraders are depleted, the "splitting" of RS into SCFAs cannot occur. This metabolic bottleneck leaves the body without its natural anti-inflammatory defenses, significantly increasing the risk of systemic inflammation. 

"Hence, diets incorporating RS-rich, fiber-dense foods, such as fruits, vegetables, and whole grains, are essential to support the proliferation of these beneficial bacteria and maintain overall gut health."

### Anaerostipes

Considered as a beneficial bacterium.
On one hand, it produces butyrate, a short-chain fatty acid essential for gut health and immune regulation. On the other hand, it can metabolize and "inactivate" the anticancer drug 5-fluorouracil (5-FU).

The study identified two distinct microbiome profiles across the lifespan (from children to older adults). Anaerostipes was consistently identified as one of the "driver taxa" for Cluster H (the healthy cluster).

## Ankylosing Spondylitis

### Introduction

https://www.usz.ch/en/disease/ankylosing-spondylitis/

Chronic inflammatory disease of the spine and sacroiliac joints. Parts stiffen with increasing inflammation due to a malfunction of the immune system, where the immune system attacks tendon insertions, leading to inflammation, pain, and eventually the potential fusion (ossification) of the vertebrae.

Triggers: While the exact cause is unknown, a combination of genetic predisposition, immune system malfunction, smoking, and potentially intestinal flora are thought to play roles.

### Link to gut microbiome

https://pmc.ncbi.nlm.nih.gov/articles/PMC9608452/

The review emphasizes a strong link between the gut and the joints, often referred to as the gut-joint axis. Approximately 70% of AS patients have subclinical intestinal inflammation, and roughly 5–10% develop clinically overt Inflammatory Bowel Disease (IBD). This suggests that the gut is a primary site where the immune response that eventually attacks the spine is first triggered.

Caused by protein, intestinal permeability weakens the gut barrier, allowing bacteria and their products to "leak" into the bloodstream, triggering a systemic immune response.

For patients with Ankylosing Spondylitis (AS), bone health is a major concern. The text explains that SCFAs:

Reduce Bone Loss: They inhibit osteoclasts (the cells that break down bone).

Metabolic Reprogramming: They force osteoclasts to change how they use energy (switching to glycolysis), which makes them less effective at destroying bone tissue.
